Algerian Forest Fires Dataset

Data Set Information:

The dataset includes 244 instances that regroup a data of two regions of Algeria,namely the Bejaia region located in the northeast of Algeria and the Sidi Bel-abbes region located in the northwet of Algeria.

122 instances for each region.

The period from June 2012 to September 2012.The dataset includes 11 attributes and 1 output attributes(class) The 244 instances have been classified into fire(138 classes) and not fire(106 classes) classes.

Attribute Information:

1.Date:(DD/MM/YYYY) Day,month('june' to 'september'),year(2012) Wather data observation.

2.Temp.tempreature noon(temperature max) in Celsius degree:22 to 42

3.RH:Relative Humidity in %:21 to 90

4.Ws:Wind speed in km/h:6 to 29

5.rain:total day in mm:0 to 16.8 FWI Components

6.Fine Fuel Moisture Code(FFMC) index from the FWI system:28.6 to 92.5

7.Duff Moisture Code(DMC) index from the FWI system:1.1 to 65.9

8.Drought Code(DC) index from the FWI systems:7 to 220.4

9.Initial Spread index(ISI) index from the FWI system:0 to 18.5

10.Buildup Index(BUI)insex from the FWI system 1.1 to 68

11.Fire Weather Index(FWI) Index:0 to 31.1

12.Classes:two classes,namely Fire and not fire.

Data Cleaning

```python
##missing values
dataset[dataset.isnull().any(axis=1)]
```

The dataset is converted into two sets based on Region from 122th index,we can make a new column based on the Region

1.:"Bejaia region Datasets"

2:"Sidi-Bel Abbes Region Dataset."

Add new column with region

```python
dataset.loc[:122,"Region"]=0
dataset.loc[122:,"Region"]=1
df=dataset
```

```python
df.info()
```

```python
df[['Region']]=df[['Region']].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
##removing the null values
df=df.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
##remove the 122nd row
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
##fix spaces in columns name
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

##Change  the required columns as integer data type

```python
df.columns
```

```python
df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']]=df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
df.info()
```

##Changing the Other columns to float data types

```python
objects = [features for features in df.columns
           if features in ['Rain', 'FFMC', 'DMC', 'DC',
                           'ISI', 'BUI', 'FWI', 'Classes']]

for i in objects:
    if i != 'Classes':
        df[i] = df[i].astype(float)

print(objects)
df.info()
```
